````python
import csv
import numpy as np

CITY = {"алматы": "Алматы", "almaty": "Алматы", "астана": "Астана", "astana": "Астана",
        "шымкент": "Шымкент", "shymkent": "Шымкент", "караганда": "Караганда",
        "актобе": "Актобе", "атырау": "Атырау"}

with open("orders.csv", encoding="utf-8") as f:
    rows = list(csv.DictReader(f))

# функции из урока 2 + превращаем колонки в массивы NumPy
amount = np.array([float(r["amount"].replace(" ", "")) for r in rows])
paid   = np.array([r["status"] == "оплачен" for r in rows])
days   = np.array([float(r["delivery_days"]) if r["delivery_days"] else np.nan for r in rows])
city   = np.array([CITY.get(r["city"].strip().lower(), r["city"].strip()) for r in rows])
cust   = np.array([r["customer_id"] for r in rows])

print(amount.shape, amount.dtype, paid.dtype)
````

## [База]
1. Найдите позицию (`argmax`) самого дорогого заказа. Какая сумма, клиент и город?
2. Выручка оплаченных заказов и сумма неоплаченных — через маску `paid` и `~paid`.
3. Сколько заказов в Шымкенте и какая у них доля от всех заказов?

````python
# База
#1
max_idx = amount.argmax()
max_val = amount[max_idx]
max_cust = cust[max_idx]
max_city = city[max_idx]

print(f"Позиция: {max_idx}")
print(f"Сумма: {max_val:,.0f}")
print(f"Клиент: {max_cust}")
print(f"Город: {max_city}")
````

````python
#2
paid_rev = amount[paid].sum()
unpaid_rev = amount[~paid].sum()

print(f"Выручка оплаченных (paid): {paid_rev:,.0f}")
print(f"Сумма неоплаченных (~paid): {unpaid_rev:,.0f}")
````

````python
# 3
shym_mask = (city == "Шымкент")
shym_count = shym_mask.sum()
shym_share = shym_mask.mean()

print(f"Заказов в Шымкенте: {shym_count}")
print(f"Доля от всех заказов: {shym_share:.4f} ({shym_share * 100:.1f}%)")
````

````python
# Стандарт
#4
mask_all = paid & (amount > 50000)
loss_all = (amount[mask_all] * 0.15).sum()

mask_clean = paid & (amount > 50000) & (cust != "TEST")
loss_clean = (amount[mask_clean] * 0.15).sum()

print(f"Недополученная выручка (со всеми заказами): {loss_all:,.0f}")
print(f"Недополученная выручка (без клиента TEST): {loss_clean:,.0f}")
print(f"Разница: в {loss_all / loss_clean:.2f} раза")
#3 фейковых заказа тестового аккаунта TEST по 9 999 990, из-за которых сумма потерь была завышена почти в 5 раз.
````

````python
#5
for c in np.unique(city):
    mask_c = (city == c) & (~np.isnan(days))
    rate = (days[mask_c] <= 2).mean()
    print(f"{c:<12}: {rate * 100:.1f}%")
````
