````python
# Данные: заказы за неделю (не менять!)
cities  = [" almaty", "Astana", "SHYMKENT", "Almaty ", "karaganda", "astana", "Almaty", "Aktobe", "Astana", "shymkent", "Almaty", "Atyrau"]
amounts = ["15990", "4500", "32000", "870", "12500", "56000", "2300", "9900", "18000", "7400", "41000", "3300"]
paid    = [True, True, False, True, True, False, True, True, True, False, True, True]
days    = ["пн", "пн", "вт", "вт", "ср", "ср", "чт", "пт", "пт", "сб", "сб", "вс"]
````

## [База]
1. Приведите все города к единому виду (`"Almaty"`, `"Astana"`, …) и сохраните в новый список `cities_clean`.
2. Преобразуйте `amounts` из строк в числа: список `amounts_int`.
3. Посчитайте: общее число заказов, общую сумму, средний чек (округлить до целого).
4. Сколько заказов не оплачено и на какую сумму?
5. Выведите таблицу в формате `город | сумма | статус`, используя `zip` и f-строку.

````python
# База
#1
cities_clean = [city.strip().title() for city in cities]
print(cities_clean)
````

````python
#2
amounts_int = [int(x) for x in amounts]
print(amounts_int)
````

````python
#3
total_orders = len(amounts_int)
total_sum = sum(amounts_int)
average_receipt = round(total_sum/total_orders)

print("общее число заказов:", total_orders)
print("общяя сумма:", total_sum)
print("средний чек:", average_receipt)
````

````python
#4
unpaid_count = 0
unpaid_sum = 0

for amt, is_paid in zip(amounts_int, paid):
    if not is_paid:
        unpaid_count += 1
        unpaid_sum += amt

print("Количество неоплаченных товаров:", unpaid_count)
print("Общая сумма неоплаченных товаров:", unpaid_sum)
````

````python
#5
for city, amt, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid:
        status = "Оплачен"
    else:
        status = "Не оплачен"
    print(f"{city:<10}|{amt:<6}|{is_paid}")
````

++## [Стандарт]
6. Выручка (только оплаченные) по Алматы и по Астане отдельно.
7. Номер (индекс) и город самого крупного **оплаченного** заказа.
8. Доля заказов выходного дня (`сб`, `вс`) в процентах.
9. Задача «копилка»: студент откладывает 15 000 ₸ в месяц, курс стоит 400 000 ₸. С помощью `while` посчитайте, через сколько месяцев он накопит. А если каждый месяц откладывать на 1 000 ₸ больше, чем в прошлый?

````python
# Стандарт
#6
rev_astana = 0
rev_almaty = 0

for city, amt, is_paid in zip(cities_clean, amounts_int, paid):
    if is_paid:
        if city == "Almaty":
            rev_almaty += amt
        elif city == "Astana":
            rev_astana += amt
print("Выручка по Астане:", rev_astana)
print("Выручка по Алматы:", rev_almaty)
````

````python
#7
max_amount = 0
max_city = ""
max_id = -1

for i, (city, amt, is_paid) in enumerate(zip(cities_clean, amounts_int, paid)):
    if is_paid and amt > max_amount:
        max_amount = amt
        max_city = city
        max_id = i
        
print(f"Индекс: {max_id}, Город: {max_city}, Сумма: {max_amount}")
````
